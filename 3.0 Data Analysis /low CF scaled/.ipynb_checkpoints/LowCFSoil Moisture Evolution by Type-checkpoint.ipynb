```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
```

```python
file = '/global/scratch/users/toireasamarie/0.1x0.1_deg_gridded_datasets/gridded_and_crop_filtered/final_paper_files/7day_events_long.parquet'
events_long = pd.read_parquet(file, engine='pyarrow')
```

```python
plot_df = events_long.copy()

#keep only the two classes you want
plot_df = plot_df[plot_df["sm_class"].isin(["Increasing", "Decreasing"])].copy()

#drop missing values needed for plotting
plot_df = plot_df.dropna(subset=["lag_days", "WFPS_c", "sm_class"])


#compute mean + SEM by lag
summary = (
    plot_df.groupby(["sm_class", "lag_days"])["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary["sem"] = summary["std"] / np.sqrt(summary["count"])

#split out the two classes
inc = summary[summary["sm_class"] == "Increasing"].copy()
dec = summary[summary["sm_class"] == "Decreasing"].copy()

#plot
fig, ax = plt.subplots(figsize=(10, 6))

#increasing
ax.plot(
    inc["lag_days"],
    inc["mean"],
    label="Increasing",
    linewidth=2)
ax.fill_between(
    inc["lag_days"],
    inc["mean"] - inc["sem"],
    inc["mean"] + inc["sem"],
    alpha=0.25)

#decreasing
ax.plot(
    dec["lag_days"],
    dec["mean"],
    label="Decreasing",
    linewidth=2)
ax.fill_between(
    dec["lag_days"],
    dec["mean"] - dec["sem"],
    dec["mean"] + dec["sem"],
    alpha=0.25)

#pulse day marker
ax.axvline(0, linestyle="--", linewidth=1.5)
ax.axvspan(-0.5, 0.5, alpha=0.12)

ax.set_xlabel("Lag (days from pulse start)")
ax.set_ylabel("Soil moisture (WFPS)")
ax.set_title("Soil moisture evolution around NO$_2$ pulses\nIncreasing vs Decreasing cases")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ── Soil type preparation (from soil_props_and_pulses notebook) ────────────
import warnings
warnings.filterwarnings("ignore", category=FutureWarning, module="seaborn")

soil_cols = [
    "pct_clay", "pct_clay_loam", "pct_loam", "pct_loamy_sand",
    "pct_sand", "pct_sandy_clay", "pct_sandy_clay_loam",
    "pct_sandy_loam", "pct_silt", "pct_silt_loam",
    "pct_silty_clay", "pct_silty_clay_loam"
]

df = events_long.copy()

# Dominant soil type per grid cell
df["dominant_soil"] = df[soil_cols].idxmax(axis=1).str.replace("pct_", "")
df["max_frac"] = df[soil_cols].max(axis=1)

# Endmember filter: >=70% single soil type, soil types with >50 events
threshold = 0.70
df_endmembers = df[df["max_frac"] >= threshold].copy()

counts = df_endmembers["dominant_soil"].value_counts()
keep_soils = counts[counts > 50].index
df_endmembers = df_endmembers[df_endmembers["dominant_soil"].isin(keep_soils)].copy()

# Broader 4-way soil group mapping
soil_group_map = {
    "clay":             "clay_rich",
    "silty_clay":       "clay_rich",
    "clay_loam":        "clay_rich",
    "silty_clay_loam":  "clay_rich",
    "loam":             "loam_rich",
    "silt_loam":        "loam_rich",
    "sandy_loam":       "sandy_loam",
    "sand":             "sand_rich",
    "loamy_sand":       "sand_rich",
}
df_endmembers["soil_group"] = df_endmembers["dominant_soil"].map(soil_group_map)

# Keep only rows with a soil group and the two sm_class values we want
df_end = df_endmembers[
    df_endmembers["soil_group"].notna() &
    df_endmembers["sm_class"].isin(["Increasing", "Decreasing"])
].dropna(subset=["lag_days", "WFPS_c"]).copy()

soil_group_order = ["clay_rich", "loam_rich", "sandy_loam", "sand_rich"]
group_palette = {
    "clay_rich":  "#c0392b",
    "loam_rich":  "#e67e22",
    "sandy_loam": "#27ae60",
    "sand_rich":  "#2980b9",
}

print("Events per soil group × sm_class:")
print(df_end.groupby(["soil_group", "sm_class"])["event_id"].nunique().unstack())
```

```python
# ── Four-panel combined figure ─────────────────────────────────────────────

fig = plt.figure(figsize=(18, 10))

# Grid: top row 3 panels (equal width), bottom row 1 panel
gs = fig.add_gridspec(2, 3, hspace=0.40, wspace=0.32,
                      height_ratios=[1.1, 0.9])

ax1 = fig.add_subplot(gs[0, :])    # top: spans all 3 columns
ax2 = fig.add_subplot(gs[1, 0])    # bottom-left:  Increasing × soil
ax3 = fig.add_subplot(gs[1, 1])    # bottom-mid:   Decreasing × soil
ax4 = fig.add_subplot(gs[1, 2])    # bottom-right: counts/pct stacked bar

# ── Colours for sm_class in panel 1 ───────────────────────────────────────
sm_colors = {"Increasing": "#4472C4", "Decreasing": "#ED7D31"}

# ── Panel 1: overall WFPS, Increasing vs Decreasing ───────────────────────
plot_df = events_long[events_long["sm_class"].isin(["Increasing", "Decreasing"])].copy()
plot_df = plot_df.dropna(subset=["lag_days", "WFPS_c", "sm_class"])

summary = (
    plot_df.groupby(["sm_class", "lag_days"])["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)
summary["sem"] = summary["std"] / np.sqrt(summary["count"])

for cls in ["Increasing", "Decreasing"]:
    sub = summary[summary["sm_class"] == cls]
    ax1.plot(sub["lag_days"], sub["mean"], label=cls,
             color=sm_colors[cls], linewidth=2)
    ax1.fill_between(sub["lag_days"],
                     sub["mean"] - sub["sem"],
                     sub["mean"] + sub["sem"],
                     color=sm_colors[cls], alpha=0.20)

ax1.axvline(0, linestyle="--", linewidth=1.5, color="gray", label="Pulse day")
ax1.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
ax1.set_xlabel("Lag (days from pulse start)", fontsize=11)
ax1.set_ylabel("WFPS (%)", fontsize=11)
ax1.set_title("(a)  Overall soil moisture evolution around NO₂ pulses", fontsize=12, loc="left")
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# ── Panels 2 & 3: Increasing / Decreasing by soil group ───────────────────
for ax, cls, label in [
    (ax2, "Increasing", "(b)  Increasing cases by soil type"),
    (ax3, "Decreasing", "(c)  Decreasing cases by soil type"),
]:
    sub_df = df_end[df_end["sm_class"] == cls]
    soil_summary = (
        sub_df.groupby(["soil_group", "lag_days"])["WFPS_c"]
        .agg(["mean", "std", "count"])
        .reset_index()
    )
    soil_summary["sem"] = soil_summary["std"] / np.sqrt(soil_summary["count"])

    for grp in soil_group_order:
        grp_sub = soil_summary[soil_summary["soil_group"] == grp]
        if grp_sub.empty:
            continue
        ax.plot(grp_sub["lag_days"], grp_sub["mean"],
                color=group_palette[grp], linewidth=2,
                label=grp.replace("_", " "))
        ax.fill_between(grp_sub["lag_days"],
                        grp_sub["mean"] - grp_sub["sem"],
                        grp_sub["mean"] + grp_sub["sem"],
                        color=group_palette[grp], alpha=0.18)

    ax.axvline(0, linestyle="--", linewidth=1.2, color="gray")
    ax.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
    ax.set_xlabel("Lag (days from pulse start)", fontsize=10)
    ax.set_ylabel("WFPS (%)", fontsize=10)
    ax.set_title(label, fontsize=11, loc="left")
    ax.legend(fontsize=8, title="Soil group", title_fontsize=8)
    ax.grid(True, alpha=0.3)

# Share y-axis between panels 2 & 3
y_min = min(ax2.get_ylim()[0], ax3.get_ylim()[0])
y_max = max(ax2.get_ylim()[1], ax3.get_ylim()[1])
ax2.set_ylim(y_min, y_max)
ax3.set_ylim(y_min, y_max)

# ── Panel 4: % and N of events per soil group × sm_class ──────────────────
counts_split = (
    df_end.groupby(["soil_group", "sm_class"])["event_id"]
    .nunique()
    .unstack()
    .reindex(soil_group_order)
    .fillna(0)
)
counts_split["total"] = counts_split.sum(axis=1)

# Fractional bars
counts_frac = counts_split[["Increasing", "Decreasing"]].div(counts_split["total"], axis=0) * 100

bar_colors = ["#4472C4", "#ED7D31"]
bottom = np.zeros(len(soil_group_order))
x = np.arange(len(soil_group_order))
bar_width = 0.55

for col, color in zip(["Increasing", "Decreasing"], bar_colors):
    vals = counts_frac[col].values
    ax4.bar(x, vals, bottom=bottom, color=color, width=bar_width,
            label=col, alpha=0.85)
    # Annotate with N inside each segment if large enough
    for xi, (v, b) in enumerate(zip(vals, bottom)):
        if v > 5:
            n = int(counts_split.loc[soil_group_order[xi], col])
            ax4.text(xi, b + v / 2, f"n={n}", ha="center", va="center",
                     fontsize=8, color="white", fontweight="bold")
    bottom += vals

ax4.set_xticks(x)
ax4.set_xticklabels([g.replace("_", "\n") for g in soil_group_order], fontsize=9)
ax4.set_ylabel("% of events", fontsize=10)
ax4.set_ylim(0, 105)
ax4.set_title("(d)  Event composition by soil group", fontsize=11, loc="left")
ax4.legend(fontsize=8, title="SM class", title_fontsize=8)
ax4.grid(True, alpha=0.3, axis="y")

# Add total N above each bar
for xi, grp in enumerate(soil_group_order):
    total = int(counts_split.loc[grp, "total"])
    ax4.text(xi, 102, f"N={total}", ha="center", va="bottom",
             fontsize=8, color="black")

plt.suptitle("Soil Moisture Evolution around NO₂ Pulses — by Soil Type",
             fontsize=14, y=1.01, fontweight="bold")

# out_path = "/global/home/users/toireasamarie/Soil_Nox_Pulses_Paper_Notebooks/SM_evolution_4panel.png"
# plt.savefig(out_path, dpi=300, bbox_inches="tight")
#print(f"Saved → {out_path}")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ── Four-panel combined figure with table in panel 4 ──────────────────────
fig = plt.figure(figsize=(18, 10))

gs = fig.add_gridspec(
    2, 3,
    hspace=0.40,
    wspace=0.32,
    height_ratios=[1.5, 1.0]  # Adjust these numbers to your preference
)

ax1 = fig.add_subplot(gs[0, :])   # top: spans all 3 columns
ax2 = fig.add_subplot(gs[1, 0])   # bottom-left
ax3 = fig.add_subplot(gs[1, 1])   # bottom-middle
ax4 = fig.add_subplot(gs[1, 2])   # bottom-right (table)

# ── Colours for sm_class in panel 1 ───────────────────────────────────────
sm_colors = {"Increasing": "#4472C4", "Decreasing": "#ED7D31"}

# ── Panel 1: overall WFPS, Increasing vs Decreasing ───────────────────────
plot_df = events_long[events_long["sm_class"].isin(["Increasing", "Decreasing"])].copy()
plot_df = plot_df.dropna(subset=["lag_days", "WFPS_c", "sm_class"])

summary = (
    plot_df.groupby(["sm_class", "lag_days"])["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)
summary["sem"] = summary["std"] / np.sqrt(summary["count"])

for cls in ["Increasing", "Decreasing"]:
    sub = summary[summary["sm_class"] == cls]
    ax1.plot(
        sub["lag_days"], sub["mean"],
        label=cls, color=sm_colors[cls], linewidth=2
    )
    ax1.fill_between(
        sub["lag_days"],
        sub["mean"] - sub["sem"],
        sub["mean"] + sub["sem"],
        color=sm_colors[cls], alpha=0.20
    )

ax1.axvline(0, linestyle="--", linewidth=1.5, color="gray", label="Pulse day")
ax1.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
ax1.set_xlabel("Lag (days from pulse start)", fontsize=11)
ax1.set_ylabel("WFPS (%)", fontsize=11)
ax1.set_title("(a)  Overall soil moisture evolution around NO$_2$ pulses", fontsize=12, loc="left")
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# ── Panels 2 & 3: Increasing / Decreasing by soil group ───────────────────
for ax, cls, label in [
    (ax2, "Increasing", "(b)  Increasing cases by soil type"),
    (ax3, "Decreasing", "(c)  Decreasing cases by soil type"),
]:
    sub_df = df_end[df_end["sm_class"] == cls].copy()

    soil_summary = (
        sub_df.groupby(["soil_group", "lag_days"])["WFPS_c"]
        .agg(["mean", "std", "count"])
        .reset_index()
    )
    soil_summary["sem"] = soil_summary["std"] / np.sqrt(soil_summary["count"])

    for grp in soil_group_order:
        grp_sub = soil_summary[soil_summary["soil_group"] == grp]
        if grp_sub.empty:
            continue

        ax.plot(
            grp_sub["lag_days"], grp_sub["mean"],
            color=group_palette[grp], linewidth=2,
            label=grp.replace("_", " ")
        )
        ax.fill_between(
            grp_sub["lag_days"],
            grp_sub["mean"] - grp_sub["sem"],
            grp_sub["mean"] + grp_sub["sem"],
            color=group_palette[grp], alpha=0.18
        )

    ax.axvline(0, linestyle="--", linewidth=1.2, color="gray")
    ax.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
    ax.set_xlabel("Lag (days from pulse start)", fontsize=10)
    ax.set_ylabel("WFPS (%)", fontsize=10)
    ax.set_title(label, fontsize=11, loc="left")
    ax.legend(fontsize=8, title="Soil group", title_fontsize=8)
    ax.grid(True, alpha=0.3)

# Match y-limits for panels 2 and 3
y_min = min(ax2.get_ylim()[0], ax3.get_ylim()[0])
y_max = max(ax2.get_ylim()[1], ax3.get_ylim()[1])
ax2.set_ylim(y_min, y_max)
ax3.set_ylim(y_min, y_max)

# ── Panel 4: table of Increasing/Decreasing counts and percentages ────────
# Use unique events so each event is counted once per soil group/sm_class
event_counts = (
    df_end.groupby(["soil_group", "sm_class"])["event_id"]
    .nunique()
    .unstack()
    .reindex(soil_group_order)
    .fillna(0)
)

# Ensure both columns exist
for col in ["Increasing", "Decreasing"]:
    if col not in event_counts.columns:
        event_counts[col] = 0

event_counts = event_counts[["Increasing", "Decreasing"]]
event_counts["Total"] = event_counts["Increasing"] + event_counts["Decreasing"]

event_counts["Inc %"] = np.where(
    event_counts["Total"] > 0,
    100 * event_counts["Increasing"] / event_counts["Total"],
    np.nan
)
event_counts["Dec %"] = np.where(
    event_counts["Total"] > 0,
    100 * event_counts["Decreasing"] / event_counts["Total"],
    np.nan
)

# Build display table
table_df = pd.DataFrame({
    "Soil group": [g.replace("_", " ").title() for g in event_counts.index],
    "Inc n": event_counts["Increasing"].astype(int).values,
    "Inc %": [f"{x:.1f}%" if pd.notnull(x) else "—" for x in event_counts["Inc %"]],
    "Dec n": event_counts["Decreasing"].astype(int).values,
    "Dec %": [f"{x:.1f}%" if pd.notnull(x) else "—" for x in event_counts["Dec %"]],
    "Total n": event_counts["Total"].astype(int).values
})

ax4.axis("off")
ax4.set_title("(d)  Event composition by soil group", fontsize=11, loc="left")

table = ax4.table(
    cellText=table_df.values,
    colLabels=table_df.columns,
    loc="center",
    cellLoc="center",
    colLoc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.15, 1.55)

# Style header + rows
for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight="bold", color="black")
        cell.set_facecolor("#D9EAF7")
    else:
        if row % 2 == 0:
            cell.set_facecolor("#F7F7F7")
        else:
            cell.set_facecolor("white")

# Optional: color the Inc/Dec header columns a little differently
table[(0, 1)].set_facecolor("#DCE6F8")  # Inc n
table[(0, 2)].set_facecolor("#DCE6F8")  # Inc %
table[(0, 3)].set_facecolor("#FBE5D6")  # Dec n
table[(0, 4)].set_facecolor("#FBE5D6")  # Dec %

plt.suptitle(
    "Soil Moisture Evolution around NO$_2$ Pulses — by Soil Type",
    fontsize=14, y=1.01, fontweight="normal"
)

# out_path = "/global/home/users/toireasamarie/Soil_Nox_Pulses_Paper_Notebooks/SM_evolution_4panel_table.png"
# plt.savefig(out_path, dpi=300, bbox_inches="tight")
# print(f"Saved → {out_path}")

plt.show()
```

```python
# ── 2x2 four-panel figure ──────────────────────────────────────────────────
fig = plt.figure(figsize=(18, 10))
# ── Four-panel combined figure ──────────────────────────────────────────
# We use 2 rows and 3 columns to allow for flexible spanning
gs = fig.add_gridspec(
    2, 3,
    hspace=0.40,
    wspace=0.35, # Slightly wider spacing to separate the table
    height_ratios=[1, 1] # Now we keep heights equal
)

# ax1 now spans columns 0 and 1 (longer/wider)
ax1 = fig.add_subplot(gs[0, 0:2])   

# ax2 and ax3 stay in the first two columns of the bottom row (equal size)
ax2 = fig.add_subplot(gs[1, 0])    
ax3 = fig.add_subplot(gs[1, 1])    

# ax4 (the table) can span the entire right-hand column (0 to 1 in row index)
# This keeps the right side dedicated to the data summary
ax4 = fig.add_subplot(gs[:, 2])
# ── Colours for sm_class in panel 1 ───────────────────────────────────────
sm_colors = {"Increasing": "#4472C4", "Decreasing": "#ED7D31"}

# ── Panel (a): overall WFPS, Increasing vs Decreasing ─────────────────────
plot_df = events_long[events_long["sm_class"].isin(["Increasing", "Decreasing"])].copy()
plot_df = plot_df.dropna(subset=["lag_days", "WFPS_c", "sm_class"])

summary = (
    plot_df.groupby(["sm_class", "lag_days"])["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)
summary["sem"] = summary["std"] / np.sqrt(summary["count"])

for cls in ["Increasing", "Decreasing"]:
    sub = summary[summary["sm_class"] == cls]
    ax1.plot(
        sub["lag_days"], sub["mean"],
        label=cls, color=sm_colors[cls], linewidth=2
    )
    ax1.fill_between(
        sub["lag_days"],
        sub["mean"] - sub["sem"],
        sub["mean"] + sub["sem"],
        color=sm_colors[cls], alpha=0.20
    )

ax1.axvline(0, linestyle="--", linewidth=1.5, color="gray", label="Pulse day")
ax1.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
ax1.set_xlabel("Lag (days from pulse start)", fontsize=11)
ax1.set_ylabel("WFPS (%)", fontsize=11)
ax1.set_title("(a) Overall soil moisture evolution around NO$_2$ pulses", fontsize=12, loc="left")
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# ── Panels (c) and (d): split by soil group ───────────────────────────────
for ax, cls, label in [
    (ax3, "Increasing", "(c) Increasing cases by soil type"),
    (ax4, "Decreasing", "(d) Decreasing cases by soil type"),
]:
    sub_df = df_end[df_end["sm_class"] == cls].copy()

    soil_summary = (
        sub_df.groupby(["soil_group", "lag_days"])["WFPS_c"]
        .agg(["mean", "std", "count"])
        .reset_index()
    )
    soil_summary["sem"] = soil_summary["std"] / np.sqrt(soil_summary["count"])

    for grp in soil_group_order:
        grp_sub = soil_summary[soil_summary["soil_group"] == grp]
        if grp_sub.empty:
            continue

        ax.plot(
            grp_sub["lag_days"], grp_sub["mean"],
            color=group_palette[grp], linewidth=2,
            label=grp.replace("_", " ")
        )
        ax.fill_between(
            grp_sub["lag_days"],
            grp_sub["mean"] - grp_sub["sem"],
            grp_sub["mean"] + grp_sub["sem"],
            color=group_palette[grp], alpha=0.18
        )

    ax.axvline(0, linestyle="--", linewidth=1.2, color="gray")
    ax.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
    ax.set_xlabel("Lag (days from pulse start)", fontsize=10)
    ax.set_ylabel("WFPS (%)", fontsize=10)
    ax.set_title(label, fontsize=11, loc="left")
    ax.legend(fontsize=8, title="Soil group", title_fontsize=8)
    ax.grid(True, alpha=0.3)

# Match y-axis for bottom two panels
y_min = min(ax3.get_ylim()[0], ax4.get_ylim()[0])
y_max = max(ax3.get_ylim()[1], ax4.get_ylim()[1])
ax3.set_ylim(y_min, y_max)
ax4.set_ylim(y_min, y_max)

# ── Panel (b): table of event composition by soil group ───────────────────
# unique events per soil_group × sm_class
event_counts = (
    df_end.groupby(["soil_group", "sm_class"])["event_id"]
    .nunique()
    .unstack()
    .reindex(soil_group_order)
    .fillna(0)
)

for col in ["Increasing", "Decreasing"]:
    if col not in event_counts.columns:
        event_counts[col] = 0

event_counts = event_counts[["Increasing", "Decreasing"]]
event_counts["Total"] = event_counts["Increasing"] + event_counts["Decreasing"]

event_counts["Inc %"] = np.where(
    event_counts["Total"] > 0,
    100 * event_counts["Increasing"] / event_counts["Total"],
    np.nan
)
event_counts["Dec %"] = np.where(
    event_counts["Total"] > 0,
    100 * event_counts["Decreasing"] / event_counts["Total"],
    np.nan
)

table_df = pd.DataFrame({
    "Soil group": [g.replace("_", " ").title() for g in event_counts.index],
    "Inc n": event_counts["Increasing"].astype(int).values,
    "Inc %": [f"{x:.1f}%" if pd.notnull(x) else "—" for x in event_counts["Inc %"]],
    "Dec n": event_counts["Decreasing"].astype(int).values,
    "Dec %": [f"{x:.1f}%" if pd.notnull(x) else "—" for x in event_counts["Dec %"]],
    "Total n": event_counts["Total"].astype(int).values
})

ax2.axis("off")
ax2.set_anchor('C')
ax2.set_title("(b) Event composition by soil group", fontsize=12, loc="left")

table = ax2.table(
    cellText=table_df.values,
    colLabels=table_df.columns,
    loc="center",
    cellLoc="center",
    colLoc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1.0, 1.5)

for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight="bold")
        cell.set_facecolor("#D9EAF7")
    else:
        cell.set_facecolor("#F7F7F7" if row % 2 == 0 else "white")

# highlight increasing / decreasing columns
table[(0, 1)].set_facecolor("#DCE6F8")
table[(0, 2)].set_facecolor("#DCE6F8")
table[(0, 3)].set_facecolor("#FBE5D6")
table[(0, 4)].set_facecolor("#FBE5D6")

plt.suptitle(
    "Soil Moisture Evolution around NO$_2$ Pulses",
    fontsize=15, y=0.98, fontweight="normal"
)

# out_path = "/global/home/users/toireasamarie/Soil_Nox_Pulses_Paper_Notebooks/SM_evolution_2x2.png"
# plt.savefig(out_path, dpi=300, bbox_inches="tight")
# print(f"Saved → {out_path}")

plt.show()
```
