```python
# ── Filter: keep only pulses detected after 10:00 AM (local time) ──────────
# 'pulse_start == True' marks pulse detection rows
# 'datetime_local' holds the local datetime; we filter on its hour

# Ensure datetime_local is parsed as datetime
df['datetime_local'] = pd.to_datetime(df['datetime_local'])

# Identify all event_ids where the pulse was detected at or after 10:00 AM
pulse_rows = df[df['pulse_start'] == True].copy()
after_10_event_ids = pulse_rows[
    pulse_rows['datetime_local'].dt.hour >= 10
]['event_id'].unique()

n_before = df['event_id'].nunique()
df = df[df['event_id'].isin(after_10_event_ids)].copy()
n_after = df['event_id'].nunique()

print(f'Events before filter : {n_before:,}')
print(f'Events after  filter : {n_after:,}  (pulse_start detected >= 10:00 local time)')
print(f'Events removed       : {n_before - n_after:,}')
```
