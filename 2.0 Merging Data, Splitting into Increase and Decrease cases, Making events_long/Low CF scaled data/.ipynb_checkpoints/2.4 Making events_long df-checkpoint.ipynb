```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
file ='/global/scratch/users/toireasamarie/0.1x0.1_deg_gridded_datasets/gridded_and_crop_filtered/final_paper_files/lowCF_scaled_filtered_merged_with_incdec.parquet'
df= pd.read_parquet(file, engine='pyarrow')
```

```python
#prep
df = df.copy()
df["datetime"] = pd.to_datetime(df["datetime"])
df["datetime_local"] = pd.to_datetime(df["datetime_local"])

# Make sure data are sorted
df = df.sort_values(["grid_lat", "grid_lon", "datetime"]).reset_index(drop=True)
```

```python
#columns to keep 
keep_cols = [
    "dominant_texture", "pct_clay", "pct_clay_loam", "pct_loam",
    "pct_loamy_sand", "pct_sand", "pct_sandy_clay", "pct_sandy_clay_loam",
    "pct_sandy_loam", "pct_silt", "pct_silt_loam", "pct_silty_clay",
    "pct_silty_clay_loam", "state_name", "datetime_local", "pre_slope", "sm_class"
]

#variables to average within each lag day 
avg_cols = [
    "no2_mean",
    'sm_surface',
    "WFPS_c",
    "h_precip_mm",
    "surface_temp",
    "soil_temp_layer1"
]
```

```python
#get pulse rows and give ID 
pulse_df = df[df["pulse_start"] == True].copy().reset_index(drop=True)

# Give each pulse a unique ID
pulse_df["event_id"] = np.arange(len(pulse_df))

records = []
```

```python
from tqdm import tqdm

# build windows with progress bar
for _, pulse in tqdm(pulse_df.iterrows(), total=len(pulse_df), desc="Building event windows"):
    lat = pulse["grid_lat"]
    lon = pulse["grid_lon"]
    t0  = pulse["datetime"]

    # all data at this grid cell
    sub = df[(df["grid_lat"] == lat) & (df["grid_lon"] == lon)].copy()

    # keep only +/- 7 day window
    sub = sub[
        (sub["datetime"] >= t0 - pd.Timedelta(days=7)) &
        (sub["datetime"] <= t0 + pd.Timedelta(days=7))
    ].copy()

    if sub.empty:
        continue

    # lag in integer days relative to pulse time
    sub["lag_days"] = ((sub["datetime"] - t0) / pd.Timedelta(days=1)).apply(np.floor).astype(int)

    # keep only exact lag bins -7 ... +7
    sub = sub[(sub["lag_days"] >= -7) & (sub["lag_days"] <= 7)].copy()

    # daily averages for the requested variables
    grouped = (
        sub.groupby("lag_days", as_index=False)[avg_cols]
        .mean()
    )

    # add pulse metadata back onto every lag row
    grouped["event_id"] = pulse["event_id"]
    grouped["grid_lat"] = lat
    grouped["grid_lon"] = lon
    grouped["datetime_event"] = t0
    grouped["pulse_date"] = t0.normalize()

    for col in keep_cols:
        grouped[col] = pulse[col]

    records.append(grouped)
```

```python
#combine into events_long
events_long = pd.concat(records, ignore_index=True)
front_cols = ["event_id", "grid_lat", "grid_lon", "datetime_event", "pulse_date", "lag_days",
              "no2_mean","WFPS_c", "sm_surface", "h_precip_mm", "surface_temp", "soil_temp_layer1"]

events_long = events_long[front_cols + keep_cols]
events_long = events_long.sort_values(["event_id", "lag_days"]).reset_index(drop=True)
events_long.head()
```

```python
plot_df = events_long.copy()

#keep only the two classes you want
plot_df = plot_df[plot_df["sm_class"].isin(["Increasing", "Decreasing"])].copy()

#drop missing values needed for plotting
plot_df = plot_df.dropna(subset=["lag_days", "WFPS_c", "sm_class"])


#compute mean + SEM by lag
summary = (
    plot_df.groupby(["sm_class", "lag_days"])["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary["sem"] = summary["std"] / np.sqrt(summary["count"])

#split out the two classes
inc = summary[summary["sm_class"] == "Increasing"].copy()
dec = summary[summary["sm_class"] == "Decreasing"].copy()

#plot
fig, ax = plt.subplots(figsize=(10, 6))

#increasing
ax.plot(
    inc["lag_days"],
    inc["mean"],
    label="Increasing",
    linewidth=2)
ax.fill_between(
    inc["lag_days"],
    inc["mean"] - inc["sem"],
    inc["mean"] + inc["sem"],
    alpha=0.25)

#decreasing
ax.plot(
    dec["lag_days"],
    dec["mean"],
    label="Decreasing",
    linewidth=2)
ax.fill_between(
    dec["lag_days"],
    dec["mean"] - dec["sem"],
    dec["mean"] + dec["sem"],
    alpha=0.25)

#pulse day marker
ax.axvline(0, linestyle="--", linewidth=1.5)
ax.axvspan(-0.5, 0.5, alpha=0.12)

ax.set_xlabel("Lag (days from pulse start)")
ax.set_ylabel("Soil moisture (WFPS)")
ax.set_title("Soil moisture evolution around NO₂ pulses\nIncreasing vs Decreasing cases")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# -----------------------------------
# 1. Prep data
# -----------------------------------
plot_df = events_long.copy()

plot_df = plot_df[plot_df["sm_class"].isin(["Increasing", "Decreasing"])].copy()
plot_df = plot_df.dropna(subset=["event_id", "lag_days", "sm_class"])

# -----------------------------------
# 2. Event-wise average first
# -----------------------------------
event_lag = (
    plot_df.groupby(["sm_class", "event_id", "lag_days"], as_index=False)[
        ["h_precip_mm", "soil_temp_layer1"]
    ]
    .mean()
)

# optional: convert soil temp to Celsius
event_lag["soil_temp_C"] = event_lag["soil_temp_layer1"] - 273.15

# lag values
lags = sorted(event_lag["lag_days"].dropna().unique())

# side-by-side positions
offset = 0.18
pos_inc = [lag - offset for lag in lags]
pos_dec = [lag + offset for lag in lags]

# helper function
def make_boxplot_data(df, value_col, class_name, lags):
    return [
        df.loc[
            (df["sm_class"] == class_name) & (df["lag_days"] == lag),
            value_col
        ].dropna().values
        for lag in lags
    ]

# build data lists
inc_precip = make_boxplot_data(event_lag, "h_precip_mm", "Increasing", lags)
dec_precip = make_boxplot_data(event_lag, "h_precip_mm", "Decreasing", lags)

inc_temp = make_boxplot_data(event_lag, "soil_temp_C", "Increasing", lags)
dec_temp = make_boxplot_data(event_lag, "soil_temp_C", "Decreasing", lags)

# -----------------------------------
# 3. Plot precipitation boxplots
# -----------------------------------
fig, ax = plt.subplots(figsize=(12, 6))

bp1 = ax.boxplot(
    inc_precip,
    positions=pos_inc,
    widths=0.32,
    patch_artist=True,
    manage_ticks=False
)

bp2 = ax.boxplot(
    dec_precip,
    positions=pos_dec,
    widths=0.32,
    patch_artist=True,
    manage_ticks=False
)

# color styling
for patch in bp1["boxes"]:
    patch.set_facecolor("steelblue")
    patch.set_alpha(0.5)

for patch in bp2["boxes"]:
    patch.set_facecolor("darkorange")
    patch.set_alpha(0.5)

ax.axvline(0, linestyle="--", linewidth=1.5, color="gray")
ax.axvspan(-0.5, 0.5, alpha=0.12, color="gray")

ax.set_xlabel("Lag (days from pulse start)")
ax.set_ylabel("Precipitation (mm)")
ax.set_title("Precipitation evolution around NO₂ pulses")
ax.set_xticks(lags)
ax.set_xticklabels(lags)
ax.grid(True, alpha=0.3)

legend_handles = [
    Patch(facecolor="steelblue", alpha=0.5, label="Increasing"),
    Patch(facecolor="darkorange", alpha=0.5, label="Decreasing")
]
ax.legend(handles=legend_handles)

plt.tight_layout()
plt.show()

# -----------------------------------
# 4. Plot soil temperature boxplots
# -----------------------------------
fig, ax = plt.subplots(figsize=(12, 6))

bp1 = ax.boxplot(
    inc_temp,
    positions=pos_inc,
    widths=0.32,
    patch_artist=True,
    manage_ticks=False
)

bp2 = ax.boxplot(
    dec_temp,
    positions=pos_dec,
    widths=0.32,
    patch_artist=True,
    manage_ticks=False
)

for patch in bp1["boxes"]:
    patch.set_facecolor("steelblue")
    patch.set_alpha(0.5)

for patch in bp2["boxes"]:
    patch.set_facecolor("darkorange")
    patch.set_alpha(0.5)

ax.axvline(0, linestyle="--", linewidth=1.5, color="gray")
ax.axvspan(-0.5, 0.5, alpha=0.12, color="gray")

ax.set_xlabel("Lag (days from pulse start)")
ax.set_ylabel("Soil temperature (°C)")
ax.set_title("Soil temperature evolution around NO₂ pulses")
ax.set_xticks(lags)
ax.set_xticklabels(lags)
ax.grid(True, alpha=0.3)
ax.legend(handles=legend_handles)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------------
# 1. Pre-pulse window
# -----------------------------------
pre_df = event_lag[event_lag["lag_days"] < 0].copy()

# -----------------------------------
# 2. Apply precipitation threshold
#    (only keep meaningful rain)
# -----------------------------------
pre_df["h_precip_mm_thresh"] = pre_df["h_precip_mm"].where(pre_df["h_precip_mm"] >= 2.5, 0)

# -----------------------------------
# 3. Sum per event
# -----------------------------------
event_precip = (
    pre_df.groupby(["soil_group", "sm_class", "event_id"])["h_precip_mm_thresh"]
    .sum()
    .reset_index()
)

# -----------------------------------
# 4. Optional: drop events with no meaningful rain at all
# -----------------------------------
event_precip = event_precip[event_precip["h_precip_mm_thresh"] > 0]

# -----------------------------------
# 5. Plot
# -----------------------------------
plt.figure(figsize=(10,6))

sns.boxplot(
    data=event_precip,
    x="soil_group",
    y="h_precip_mm_thresh",
    hue="sm_class"
)

plt.ylabel("Total pre-pulse precip (mm, ≥2.5 mm only)")
plt.title("Significant precipitation by soil type and SM behavior")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plot_df = events_long.copy()
plot_df = plot_df[plot_df["sm_class"].isin(["Increasing", "Decreasing"])].copy()

# total pre-pulse precip
pre_totals = (
    plot_df[plot_df["lag_days"] < 0]
    .groupby("event_id", as_index=False)
    .agg(total_precip_pre=("h_precip_mm", lambda x: x.sum(min_count=1)))
)

# manual bins
bins = [-0.01, 2.5, 5, 10, np.inf]
labels = ["0–2.5 mm", "2.5–5 mm", "5–10 mm", ">10 mm"]

pre_totals["precip_bin"] = pd.cut(
    pre_totals["total_precip_pre"],
    bins=bins,
    labels=labels
)

# merge back
plot_df = plot_df.merge(
    pre_totals[["event_id", "total_precip_pre", "precip_bin"]],
    on="event_id",
    how="left"
)

# event-wise average first
event_lag = (
    plot_df.groupby(
        ["precip_bin", "sm_class", "event_id", "lag_days"],
        observed=True
    )["sm_surface"]
    .mean()
    .reset_index()
)
# summarize across events
summary = (
    event_lag.groupby(["precip_bin", "sm_class", "lag_days"])["sm_surface"]
    .agg(["mean", "std", "count"])
    .reset_index()
)
summary["sem"] = summary["std"] / np.sqrt(summary["count"])

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)
axes = axes.flatten()

for ax, bin_label in zip(axes, labels):
    sub = summary[summary["precip_bin"] == bin_label]

    inc = sub[sub["sm_class"] == "Increasing"]
    dec = sub[sub["sm_class"] == "Decreasing"]

    ax.plot(inc["lag_days"], inc["mean"], label="Increasing", linewidth=2)
    ax.fill_between(
        inc["lag_days"],
        inc["mean"] - inc["sem"],
        inc["mean"] + inc["sem"],
        alpha=0.25
    )

    ax.plot(dec["lag_days"], dec["mean"], label="Decreasing", linewidth=2)
    ax.fill_between(
        dec["lag_days"],
        dec["mean"] - dec["sem"],
        dec["mean"] + dec["sem"],
        alpha=0.25
    )

    ax.axvline(0, linestyle="--", linewidth=1.2, color="gray")
    ax.axvspan(-0.5, 0.5, alpha=0.10, color="gray")
    ax.set_title(f"Total pre-pulse precip: {bin_label}")
    ax.grid(True, alpha=0.3)

for ax in axes[2:]:
    ax.set_xlabel("Lag (days from pulse start)")
for ax in axes[::2]:
    ax.set_ylabel("Soil moisture (sm_surface)")

axes[0].legend()
plt.suptitle("Soil moisture evolution split by total pre-pulse precipitation", y=1.02)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plot_df = events_long.copy()
plot_df = plot_df[plot_df["sm_class"].isin(["Increasing", "Decreasing"])].copy()

texture_map = {
    "sand": "coarse",
    "loamy sand": "coarse",
    "sandy loam": "moderately_coarse",
    "loam": "medium",
    "silt loam": "medium",
    "silt": "medium",
    "clay loam": "fine",
    "silty clay loam": "fine",
    "sandy clay loam": "fine",
    "clay": "very_fine",
    "silty clay": "very_fine",
    "sandy clay": "very_fine"
}

plot_df["soil_group"] = plot_df["dominant_texture"].map(texture_map)

# total pre-pulse precip
pre_totals = (
    plot_df[plot_df["lag_days"] < 0]
    .groupby("event_id", as_index=False)
    .agg(total_precip_pre=("h_precip_mm", lambda x: x.sum(min_count=1)))
)

# manual bins
bins = [-0.01, 2.5, 5, 10, np.inf]
labels = ["0–2.5 mm", "2.5–5 mm", "5–10 mm", ">10 mm"]

pre_totals["precip_bin"] = pd.cut(
    pre_totals["total_precip_pre"],
    bins=bins,
    labels=labels
)

# merge back
plot_df = plot_df.merge(
    pre_totals[["event_id", "total_precip_pre", "precip_bin"]],
    on="event_id",
    how="left"
)

plot_df = plot_df.dropna(subset=["soil_group", "precip_bin", "sm_surface"])
```

```python
soil_order = ["coarse", "moderately_coarse", "medium", "fine", "very_fine"]

for soil in soil_order:
    soil_df = plot_df[plot_df["soil_group"] == soil].copy()

    # skip empty groups
    if soil_df.empty:
        continue

    event_lag = (
        soil_df.groupby(
            ["precip_bin", "sm_class", "event_id", "lag_days"],
            observed=True
        )["sm_surface"]
        .mean()
        .reset_index()
    )

    summary = (
        event_lag.groupby(
            ["precip_bin", "sm_class", "lag_days"],
            observed=True
        )["sm_surface"]
        .agg(["mean", "std", "count"])
        .reset_index()
    )

    summary["sem"] = summary["std"] / np.sqrt(summary["count"])

    fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)
    axes = axes.flatten()

    for ax, bin_label in zip(axes, labels):
        sub = summary[summary["precip_bin"] == bin_label]

        inc = sub[sub["sm_class"] == "Increasing"]
        dec = sub[sub["sm_class"] == "Decreasing"]

        if not inc.empty:
            ax.plot(inc["lag_days"], inc["mean"], label="Increasing", linewidth=2)
            ax.fill_between(
                inc["lag_days"],
                inc["mean"] - inc["sem"],
                inc["mean"] + inc["sem"],
                alpha=0.25
            )

        if not dec.empty:
            ax.plot(dec["lag_days"], dec["mean"], label="Decreasing", linewidth=2)
            ax.fill_between(
                dec["lag_days"],
                dec["mean"] - dec["sem"],
                dec["mean"] + dec["sem"],
                alpha=0.25
            )

        ax.axvline(0, linestyle="--", linewidth=1.2, color="gray")
        ax.axvspan(-0.5, 0.5, alpha=0.10, color="gray")
        ax.set_title(f"Total pre-pulse precip: {bin_label}")
        ax.grid(True, alpha=0.3)

    for ax in axes[2:]:
        ax.set_xlabel("Lag (days from pulse start)")
    for ax in axes[::2]:
        ax.set_ylabel("Soil moisture (sm_surface)")

    axes[0].legend()
    plt.suptitle(f"Soil moisture evolution by precip bin\nSoil group: {soil}", y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

soil_order = ["coarse", "moderately_coarse", "medium", "fine", "very_fine"]

# -------------------------------------------------
# 1. SM summary
# -------------------------------------------------
event_lag_sm = (
    plot_df.groupby(
        ["soil_group", "precip_bin", "sm_class", "event_id", "lag_days"],
        observed=True
    )["WFPS_c"]
    .mean()
    .reset_index()
)

summary_sm = (
    event_lag_sm.groupby(
        ["soil_group", "precip_bin", "sm_class", "lag_days"],
        observed=True
    )["WFPS_c"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary_sm["sem"] = summary_sm["std"] / np.sqrt(summary_sm["count"])

# -------------------------------------------------
# 2. Precip summary
#    use sum per event-day, not mean
# -------------------------------------------------
event_lag_pr = (
    plot_df.groupby(
        ["soil_group", "precip_bin", "sm_class", "event_id", "lag_days"],
        observed=True
    )["h_precip_mm"]
    .sum(min_count=1)
    .reset_index()
)

summary_pr = (
    event_lag_pr.groupby(
        ["soil_group", "precip_bin", "sm_class", "lag_days"],
        observed=True
    )["h_precip_mm"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary_pr["sem"] = summary_pr["std"] / np.sqrt(summary_pr["count"])

# -------------------------------------------------
# 3. Plot
# -------------------------------------------------
fig, axes = plt.subplots(len(soil_order), len(labels), figsize=(20, 20), sharex=True, sharey=True)

for i, soil in enumerate(soil_order):
    for j, bin_label in enumerate(labels):
        ax = axes[i, j]
        ax2 = ax.twinx()

        # -----------------------------
        # Soil moisture data
        # -----------------------------
        sub_sm = summary_sm[
            (summary_sm["soil_group"] == soil) &
            (summary_sm["precip_bin"] == bin_label)
        ]

        inc_sm = sub_sm[sub_sm["sm_class"] == "Increasing"]
        dec_sm = sub_sm[sub_sm["sm_class"] == "Decreasing"]

        # -----------------------------
        # Precipitation data
        # -----------------------------
        sub_pr = summary_pr[
            (summary_pr["soil_group"] == soil) &
            (summary_pr["precip_bin"] == bin_label)
        ]

        inc_pr = sub_pr[sub_pr["sm_class"] == "Increasing"]
        dec_pr = sub_pr[sub_pr["sm_class"] == "Decreasing"]

        # -----------------------------
        # Plot SM on left axis
        # -----------------------------
        if not inc_sm.empty:
            ax.plot(
                inc_sm["lag_days"], inc_sm["mean"],
                linewidth=1.8, label="Increasing SM"
            )
            ax.fill_between(
                inc_sm["lag_days"],
                inc_sm["mean"] - inc_sm["sem"],
                inc_sm["mean"] + inc_sm["sem"],
                alpha=0.20
            )

        if not dec_sm.empty:
            ax.plot(
                dec_sm["lag_days"], dec_sm["mean"],
                linewidth=1.8, label="Decreasing SM"
            )
            ax.fill_between(
                dec_sm["lag_days"],
                dec_sm["mean"] - dec_sm["sem"],
                dec_sm["mean"] + dec_sm["sem"],
                alpha=0.20
            )

        # -----------------------------
        # Plot precip on right axis
        # use dashed lines so it reads differently
        # -----------------------------
        if not inc_pr.empty:
            ax2.plot(
                inc_pr["lag_days"], inc_pr["mean"],
                linestyle="--", linewidth=1.5, alpha=0.9,
                label="Increasing precip"
            )

        if not dec_pr.empty:
            ax2.plot(
                dec_pr["lag_days"], dec_pr["mean"],
                linestyle="--", linewidth=1.5, alpha=0.9,
                label="Decreasing precip"
            )

        # -----------------------------
        # Formatting
        # -----------------------------
        ax.axvline(0, linestyle="--", linewidth=1, color="gray")
        ax.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
        ax.grid(True, alpha=0.25)

        if i == 0:
            ax.set_title(bin_label)
        if j == 0:
            ax.set_ylabel(f"{soil}\n Soil Moisture (WFPS)")

        # only label right axis on far right to reduce clutter
        if j == len(labels) - 1:
            ax2.set_ylabel("Daily precip total (mm)")

# x labels only on bottom row
for ax in axes[-1, :]:
    ax.set_xlabel("Lag (days from pulse start)")

# one legend from first panel
lines1, labels1 = axes[0, 0].get_legend_handles_labels()
lines2, labels2 = axes[0, 0].twinx().get_legend_handles_labels()  # won't capture plotted twin lines correctly
```

```python
# -------------------------------------------------
# 3. NO2 summary
# -------------------------------------------------
event_lag_no2 = (
    plot_df.groupby(
        ["soil_group", "precip_bin", "sm_class", "event_id", "lag_days"],
        observed=True
    )["no2_mean"]
    .mean()
    .reset_index()
)

summary_no2 = (
    event_lag_no2.groupby(
        ["soil_group", "precip_bin", "sm_class", "lag_days"],
        observed=True
    )["no2_mean"]
    .agg(["mean", "std", "count"])
    .reset_index()
)

summary_no2["sem"] = summary_no2["std"] / np.sqrt(summary_no2["count"])
```

```python
# Normalize WFPS
summary_sm["wfps_norm"] = (
    summary_sm.groupby(["soil_group", "precip_bin", "sm_class"], observed=True)["mean"]
    .transform(lambda x: (x - x.mean()) / x.std() if x.std() > 0 else x - x.mean())
)

# Normalize NO2
summary_no2["no2_norm"] = (
    summary_no2.groupby(["soil_group", "precip_bin", "sm_class"], observed=True)["mean"]
    .transform(lambda x: (x - x.mean()) / x.std() if x.std() > 0 else x - x.mean())
)
```

```python
# WFPS relative to lag 0
summary_sm["wfps_rel0"] = summary_sm.groupby(
    ["soil_group", "precip_bin", "sm_class"], observed=True
)["mean"].transform(lambda x: x)

for (soil, pbin, sclass), idx in summary_sm.groupby(
    ["soil_group", "precip_bin", "sm_class"], observed=True
).groups.items():
    sub = summary_sm.loc[idx]
    day0 = sub.loc[sub["lag_days"] == 0, "mean"]
    if len(day0) > 0 and pd.notna(day0.iloc[0]) and day0.iloc[0] != 0:
        summary_sm.loc[idx, "wfps_rel0"] = sub["mean"] / day0.iloc[0]
    else:
        summary_sm.loc[idx, "wfps_rel0"] = np.nan

# NO2 relative to lag 0
summary_no2["no2_rel0"] = summary_no2.groupby(
    ["soil_group", "precip_bin", "sm_class"], observed=True
)["mean"].transform(lambda x: x)

for (soil, pbin, sclass), idx in summary_no2.groupby(
    ["soil_group", "precip_bin", "sm_class"], observed=True
).groups.items():
    sub = summary_no2.loc[idx]
    day0 = sub.loc[sub["lag_days"] == 0, "mean"]
    if len(day0) > 0 and pd.notna(day0.iloc[0]) and day0.iloc[0] != 0:
        summary_no2.loc[idx, "no2_rel0"] = sub["mean"] / day0.iloc[0]
    else:
        summary_no2.loc[idx, "no2_rel0"] = np.nan
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

soil_order = ["coarse", "moderately_coarse", "medium", "fine", "very_fine"]

fig, axes = plt.subplots(len(soil_order), len(labels), figsize=(20, 20), sharex=True, sharey=True)

precip_ymax = summary_pr["mean"].max() * 1.1

for i, soil in enumerate(soil_order):
    for j, bin_label in enumerate(labels):
        ax = axes[i, j]
        ax2 = ax.twinx()

        # -----------------------------
        # Subset data
        # -----------------------------
        sub_sm = summary_sm[
            (summary_sm["soil_group"] == soil) &
            (summary_sm["precip_bin"] == bin_label)
        ]
        sub_pr = summary_pr[
            (summary_pr["soil_group"] == soil) &
            (summary_pr["precip_bin"] == bin_label)
        ]
        sub_no2 = summary_no2[
            (summary_no2["soil_group"] == soil) &
            (summary_no2["precip_bin"] == bin_label)
        ]

        inc_sm = sub_sm[sub_sm["sm_class"] == "Increasing"]
        dec_sm = sub_sm[sub_sm["sm_class"] == "Decreasing"]

        inc_pr = sub_pr[sub_pr["sm_class"] == "Increasing"]
        dec_pr = sub_pr[sub_pr["sm_class"] == "Decreasing"]

        inc_no2 = sub_no2[sub_no2["sm_class"] == "Increasing"]
        dec_no2 = sub_no2[sub_no2["sm_class"] == "Decreasing"]

        # -----------------------------
        # Left axis: WFPS and NO2 (normalized)
        # -----------------------------
        if not inc_sm.empty:
            ax.plot(
                inc_sm["lag_days"], inc_sm["wfps_rel0"],
                linewidth=1.8, label="Increasing WFPS"
            )

        if not dec_sm.empty:
            ax.plot(
                dec_sm["lag_days"], dec_sm["wfps_rel0"],
                linewidth=1.8, label="Decreasing WFPS"
            )

        if not inc_no2.empty:
            ax.plot(
                inc_no2["lag_days"], inc_no2["no2_rel0"],
                linestyle=":", linewidth=1.8, label="Increasing NO2"
            )

        if not dec_no2.empty:
            ax.plot(
                dec_no2["lag_days"], dec_no2["no2_rel0"],
                linestyle=":", linewidth=1.8, label="Decreasing NO2"
            )

        # -----------------------------
        # Right axis: precip
        # -----------------------------
        if not inc_pr.empty:
            ax2.plot(
                inc_pr["lag_days"], inc_pr["mean"],
                linestyle="--", linewidth=1.3, alpha=0.8, label="Increasing precip"
            )

        if not dec_pr.empty:
            ax2.plot(
                dec_pr["lag_days"], dec_pr["mean"],
                linestyle="--", linewidth=1.3, alpha=0.8, label="Decreasing precip"
            )

        ax2.set_ylim(0, precip_ymax)

        # -----------------------------
        # Formatting
        # -----------------------------
        ax.axvline(0, linestyle="--", linewidth=1, color="gray")
        ax.axvspan(-0.5, 0.5, alpha=0.08, color="gray")
        ax.grid(True, alpha=0.25)

        if i == 0:
            ax.set_title(bin_label)
        if j == 0:
            ax.set_ylabel(f"{soil}\nRelative WFPS / NO2")
        if j == len(labels) - 1:
            ax2.set_ylabel("Daily precip total (mm)")

for ax in axes[-1, :]:
    ax.set_xlabel("Lag (days from pulse start)")

legend_handles = [
    Line2D([0], [0], linewidth=2, label="Increasing WFPS"),
    Line2D([0], [0], linewidth=2, label="Decreasing WFPS"),
    Line2D([0], [0], linestyle=":", linewidth=2, label="Increasing NO2"),
    Line2D([0], [0], linestyle=":", linewidth=2, label="Decreasing NO2"),
    Line2D([0], [0], linestyle="--", linewidth=1.5, label="Increasing precip"),
    Line2D([0], [0], linestyle="--", linewidth=1.5, label="Decreasing precip"),
]

axes[0, 0].legend(handles=legend_handles, loc="upper left", fontsize=8)

plt.suptitle("WFPS, NO2, and precipitation evolution by soil group and pre-pulse precipitation", y=0.995)
plt.tight_layout()
plt.show()
```
