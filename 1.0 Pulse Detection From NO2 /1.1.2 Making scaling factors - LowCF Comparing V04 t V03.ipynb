```python
from timezonefinder import TimezoneFinder
import pytz
```

```python
tf = TimezoneFinder()

#storing timezones per unique lat/lon pair
unique_coords =v3_valid[['grid_lat', 'grid_lon']].drop_duplicates()
coord_to_tz = {}

for _, row in unique_coords.iterrows():
    tz_name = tf.timezone_at(lat=row['grid_lat'], lng=row['grid_lon'])
    coord_to_tz[(row['grid_lat'], row['grid_lon'])] = tz_name

#mapping timezone name to each row
v3_valid['timezone'] = v3_valid.apply(lambda r: coord_to_tz[(r['grid_lat'], r['grid_lon'])], axis=1)

#convert the UTC datetime to local time and extract the hour
def to_local(row):
    tz = pytz.timezone(row['timezone'])
    utc_dt = row['datetime']
    if utc_dt.tzinfo is None:
        utc_dt = pytz.utc.localize(utc_dt)
    return utc_dt.astimezone(tz)

v3_valid['datetime_local'] =v3_valid.apply(to_local, axis=1)
v3_valid['hour_local'] =v3_valid['datetime_local'].apply(lambda x: x.hour)

v3_valid.drop(columns=['timezone'], inplace=True)
print(v3_valid[['datetime', 'grid_lat', 'grid_lon', 'datetime_local', 'hour_local']].head(10))
```

```python
from matplotlib.lines import Line2D
from matplotlib.ticker import FuncFormatter, MaxNLocator

hours = list(range(7, 18))
plot_data = merged.dropna(subset=["no2_mean_v3", "no2_mean_v4"]).copy()

# Set shared axis limits using upper quantile to avoid extreme outliers dominating
upper_limit = max(plot_data["no2_mean_v3"].quantile(0.9995),
                  plot_data["no2_mean_v4"].quantile(0.9995))

# Round upward to a clean interval
upper_limit = np.ceil(upper_limit / 5e15) * 5e15
lims = (0, upper_limit)

#format ticks in units of 1e16
scale = 1e16
axis_formatter = FuncFormatter(lambda value, position: f"{value / scale:g}")

#figure setup
fig, axes = plt.subplots(2, 6, figsize=(17.8 * cm, 9 * cm), sharex=True, sharey=True)
axes = axes.flatten()

r2_dict = {}

#plotting each hour
for i, h in enumerate(hours):
    ax = axes[i]

    subset = merged.loc[merged["hour"] == h].dropna(subset=["no2_mean_v3", "no2_mean_v4"])
    pulses = subset.loc[subset["pulse_start"].fillna(False)]

    #R2
    if len(subset) > 2:
        r = subset["no2_mean_v3"].corr(subset["no2_mean_v4"])
        r2 = r**2
    else:
        r2 = np.nan

    r2_dict[h] = r2

    #obs
    ax.scatter(subset["no2_mean_v3"], subset["no2_mean_v4"],
               s=2.5, color="0.55", alpha=0.16, edgecolors="none", rasterized=True, zorder=1)

    #pulses
    ax.scatter(pulses["no2_mean_v3"], pulses["no2_mean_v4"],
               s=5, color="red", alpha=0.80, edgecolors="none",rasterized=True,zorder=2)

    #1:1 line
    ax.plot(lims, lims, linestyle=(0, (4, 2)), color="0.2", linewidth=0.8, zorder=0)

    #title
    ax.set_title(f"{h:02d}:00", pad=2)
    annotation = (f"$n$ = {len(subset):,}\n"
                  f"$n_{{pulse}}$ = {len(pulses):,}\n"
                  f"$r^2$ = {r2:.2f}")

    ax.text(0.045, 0.94, annotation,transform=ax.transAxes, 
            ha="left", va="top", fontsize=6.2,linespacing=1.05)

    ax.set_xlim(lims)
    ax.set_ylim(lims)
    ax.set_aspect("equal", adjustable="box")

    ax.xaxis.set_major_formatter(axis_formatter)
    ax.yaxis.set_major_formatter(axis_formatter)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=4))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=4))

#using the last empty panel for the legend instead 
legend_ax = axes[-1]
legend_ax.axis("off")

legend_handles = [Line2D([0], [0], marker="o", linestyle="none",
                         markerfacecolor="0.55", markeredgecolor="none", alpha=0.45, markersize=4,
                         label="All observations"),
                  Line2D([0], [0], marker="o", linestyle="none", 
                         markerfacecolor="red", markeredgecolor="none", markersize=5,
                         label="Pulse starts"),
                  Line2D([0], [0], color="0.2", linestyle=(0, (4, 2)), linewidth=0.8, 
                         label="1:1 line")]

legend_ax.legend(handles=legend_handles, loc="center left", bbox_to_anchor=(0.02, 0.50),
                 frameon=False, handlelength=2.1,handletextpad=0.7,labelspacing=0.55, borderaxespad=0)

#labels
fig.supxlabel("TEMPO V03 tropospheric NO$_2$ column\n"
              r"($10^{16}$ molecules cm$^{-2}$)",
              x=0.47, y=0.045, fontsize=8, linespacing=1.15, ha="center", multialignment="center")

fig.suptitle("TEMPO V03 versus V04 tropospheric NO$_2$ columns at each local hour", fontsize=9,y=0.985)

fig.supylabel("TEMPO V04 tropospheric NO$_2$ column\n"
              r"($10^{16}$ molecules cm$^{-2}$)",
              x=0.025, fontsize=8, linespacing=1.15, ha="center", va="center", multialignment="center")

fig.subplots_adjust(left=0.11, right=0.985,
                    bottom=0.17, top=0.9,
                    wspace=0.16, hspace=0.12)
fig.savefig("TEMPO_V03_V04_hourly_comparison_PNAS_SI.pdf", format="pdf", bbox_inches="tight", dpi=600)
plt.show()
```

# creating scaling factors for each hour

scaling factors are made by OLS through origin per hour 

```python
from scipy import stats
```

```python
hours = list(range(7, 19))
records =[]

for h in hours:
    sub = merged[merged["hour"] ==h].dropna(subset=["no2_mean_v3", "no2_mean_v4"])
    
    if len(sub) < 5:
        continue

    x =sub["no2_mean_v3"].values
    y =sub["no2_mean_v4"].values

    #least squares slope forced through origin
    slope_origin = np.dot(x, y) / np.dot(x, x)

    records.append({"hour": h, "n": len(sub), "slope_origin": slope_origin})

scaling_df = pd.DataFrame(records)
print(scaling_df.to_string(index=False, float_format="{:.4f}".format))
```

```python
scaling_df
```

```python
plt.figure(figsize=(7, 4))

plt.plot(scaling_df["hour"],scaling_df["slope_origin"],"o-")
plt.axhline(1, color="red", linestyle="--", linewidth=1)

plt.xlabel("Local hour")
plt.ylabel("OLS slope through origin")
plt.title("per-hour V3 to V4 scaling factors")
plt.xticks(scaling_df["hour"])
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
#applying per hour scaling to V3
scale_map = dict(zip(scaling_df["hour"], scaling_df["slope_origin"]))
merged["no2_v3_scaled"] = merged["no2_mean_v3"] * merged["hour"].map(scale_map)
```

```python
#plotting V4 vs scaled V3 by hour again
lims_scaled = [min(merged["no2_v3_scaled"].min(), merged["no2_mean_v4"].min()),
               max(merged["no2_v3_scaled"].max(), merged["no2_mean_v4"].max())]

fig,axes = plt.subplots(2, 6, figsize=(18, 6),sharex=True,sharey=True)
axes =axes.flatten()

for i, h in enumerate(hours):
    ax = axes[i]

    subset = merged[merged["hour"] ==h].dropna(subset=["no2_v3_scaled", "no2_mean_v4"])
    sf =scale_map.get(h, np.nan)

    ax.scatter(subset["no2_v3_scaled"],subset["no2_mean_v4"], alpha=0.15, s=3)
    ax.plot(lims_scaled, lims_scaled, "r--",linewidth=1)

    if len(subset) > 1:
        r =np.corrcoef(subset["no2_v3_scaled"], subset["no2_mean_v4"])[0, 1]
        r2 =r**2
    else:
        r2 = np.nan

    ax.set_title(f"{h:02d}:00  scale={sf:.3f}\nn={len(subset):,}  R$^2$={r2:.2f}", fontsize=8.5)
    ax.set_xlim(lims_scaled)
    ax.set_ylim(lims_scaled)

for ax in axes[6:]:
    ax.set_xlabel("V3 scaled")

for ax in axes[::6]:
    ax.set_ylabel("V4")

fig.suptitle("Midwest V4 vs scaled V3 NO$_2$\nV3 multiplied by per-hour 0-origin slope", fontsize=14)
plt.tight_layout()
plt.show()
```
