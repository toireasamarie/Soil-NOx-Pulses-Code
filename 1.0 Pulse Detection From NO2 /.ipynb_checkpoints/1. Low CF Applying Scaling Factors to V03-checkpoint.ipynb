```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
hour_scale_file ='/global/scratch/users/toireasamarie/Scaling Factors/scaling_by_hour.parquet'
state_hour_scale_file = '/global/scratch/users/toireasamarie/Scaling Factors/scaling_by_state_and_hour.parquet'
v03_file ='/global/scratch/users/toireasamarie/0.1x0.1_deg_gridded_datasets/gridded_and_crop_filtered/TEMPO_03-07_0.1x0.1GCF_all_hours.parquet'


hour_scale = pd.read_parquet(hour_scale_file, engine='pyarrow')
state_hour_scale = pd.read_parquet(state_hour_scale_file, engine='pyarrow')
v03_df = pd.read_parquet(v03_file, engine='pyarrow')

print('hour_scale shape:', hour_scale.shape)
print('state_hour_scale shape:', state_hour_scale.shape)
print('v03_df shape:', v03_df.shape)
```

# adding timezone

```python
from timezonefinder import TimezoneFinder
from zoneinfo import ZoneInfo
import numpy as np

tf = TimezoneFinder()

#tep 1: Get timezone for each unique grid cell 
unique_coords = v03_df[['grid_lat', 'grid_lon']].drop_duplicates().copy()
unique_coords['tz_name'] = unique_coords.apply(
    lambda row: tf.timezone_at(lat=row['grid_lat'], lng=row['grid_lon']),
    axis=1
)
print(f"Unique grid cells: {len(unique_coords)}")
print(unique_coords['tz_name'].value_counts())

#step 2: Merge tz_name onto v03_df 
v03_df = v03_df.merge(unique_coords, on=['grid_lat', 'grid_lon'], how='left')

# ── Step 3: Build UTC datetime 
v03_df['datetime_utc'] = pd.to_datetime(
    v03_df['date'].astype(str) + ' ' + v03_df['hour'].astype(str)
)

#step 4: Vectorized local time conversion — group by timezone 
# Much faster than row-wise apply — converts all rows sharing a timezone at once
local_times = pd.Series(index=v03_df.index, dtype='datetime64[ns]')

for tz_name, group_idx in v03_df.groupby('tz_name').groups.items():
    tz = ZoneInfo(tz_name)
    utc_times = v03_df.loc[group_idx, 'datetime_utc'].dt.tz_localize('UTC')
    local_times.loc[group_idx] = utc_times.dt.tz_convert(tz).dt.tz_localize(None)

v03_df['datetime_local'] = local_times
v03_df['local_hour'] = v03_df['datetime_local'].dt.hour.astype('int64')

print(v03_df[['datetime_utc', 'grid_lat', 'grid_lon', 'tz_name', 
              'datetime_local', 'local_hour']].head(10).to_string(index=False))
```

# adding state

```python
import geopandas as gpd
from shapely.geometry import Point

#load shapefile
states = gpd.read_file('/global/scratch/users/toireasamarie/US_State_Shapefile/tl_2024_us_state.shp')[['NAME', 'geometry']].rename(columns={'NAME': 'state_name'})
states = states.to_crs('EPSG:4326')

#get unique grid cells and convert to geodf 
unique_coords = v03_df[['grid_lat', 'grid_lon']].drop_duplicates().copy()
unique_coords['geometry'] = unique_coords.apply(
    lambda row: Point(row['grid_lon'], row['grid_lat']), axis=1
)
coords_gdf = gpd.GeoDataFrame(unique_coords, crs='EPSG:4326')

#spatial join 
coords_with_state = gpd.sjoin(coords_gdf, states, how='left', predicate='within')[
    ['grid_lat', 'grid_lon', 'state_name']
]

#merge back onto full df
v03_df = v03_df.merge(coords_with_state, on=['grid_lat', 'grid_lon'], how='left')

print(v03_df['state_name'].value_counts())
print(f"\nGrid cells with no state assigned: {v03_df['state_name'].isna().sum()}")
```

# Apply scaling

```python
#get local hour
v03_df['local_hour'] = pd.to_datetime(v03_df['datetime_local']).dt.hour.astype('int64')

#maje sure int64
hour_scale['hour']= hour_scale['hour'].astype('int64')
state_hour_scale['hour'] = state_hour_scale['hour'].astype('int64')
```

```python
v03_df.columns
```

```python
#1. Domain-wide hour scaling 
hour_scale_slim = hour_scale[['hour', 'slope_origin']].rename(
    columns={'hour': 'local_hour', 'slope_origin': 'hour_scale_factor'})

v03_df = v03_df.merge(hour_scale_slim, on='local_hour', how='left')
v03_df['no2_scaled_by_hour'] = v03_df['no2_mean'] * v03_df['hour_scale_factor']

print('Hour scaling done.')
print('Rows with no2_mean:', v03_df['no2_mean'].notna().sum())
print('Rows with no2_scaled_by_hour:', v03_df['no2_scaled_by_hour'].notna().sum())
```

```python
#2. State x hour scaling 
state_hour_slim = state_hour_scale[['state', 'hour', 'slope_origin']].rename(
    columns={'state': 'state_name', 'hour': 'local_hour', 'slope_origin': 'state_hour_scale_factor'}
)

v03_df = v03_df.merge(state_hour_slim, on=['state_name', 'local_hour'], how='left')
v03_df['no2_scaled_by_state_hour'] = v03_df['no2_mean'] * v03_df['state_hour_scale_factor']

print('State x hour scaling done.')
print('  Rows with no2_scaled_by_state_hour:', v03_df['no2_scaled_by_state_hour'].notna().sum())
```

```python
# Only flag rows that are within TEMPO hours but still missing a factor
tempo_hours = list(range(7, 19))
missing_states = (
    v03_df[
        v03_df['no2_mean'].notna() &
        v03_df['local_hour'].isin(tempo_hours) &
        v03_df['state_hour_scale_factor'].isna() &
        v03_df['state_name'].notna()
    ]['state_name'].unique()
)
if len(missing_states) > 0:
    print(f'Warning: missing factors for in-hours states: {missing_states}')
else:
    print('All in-hours, in-domain rows matched successfully.')
```

```python
#sanity check: sample output 
print(v03_df[['datetime_local', 'state_name', 'local_hour', 'no2_mean',
              'hour_scale_factor', 'no2_scaled_by_hour',
              'state_hour_scale_factor', 'no2_scaled_by_state_hour']]
      .dropna(subset=['no2_mean'])
      .head(20)
      .to_string(index=False))
```
